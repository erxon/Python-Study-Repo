# HOW TO TOKENIZE TEXT IN PYTHON

## Using `.split()`

```python
text = "how to tokenize text in python"
tokens = text.split()
print(tokens)
```

## NLTK's `word_tokenize()`

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')
text = "How to tokenize text in python using word_tokenize, and change the world?"
tokens = word_tokenize(text)
print(tokens, len(tokens))
```

## Using `re.findall()`

```python
import re

text = "How to be a good person?"
tokens = re.findall(r'\w+', text)
print(tokens)
```

## Using `str.split()` in Pandas dataframe

```python
import pandas as pd

df = pd.DataFrame({"text": ["How to use Pandas dataframe for text tokenizing?"]})
df['tokens'] = df['text'].str.split()
print(df['tokens'][0])
```

## Using Gensim's `tokenize()` function

```python
from gensim.utils import tokenize

text = "How to use gensim to tokenize text"
tokens = list(tokenize(text))
print(tokens)
```
